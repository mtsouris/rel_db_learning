## Step 5: See your tables

Your database now has four empty tables. Let's confirm by listing them.
The following query reads PostgreSQL's catalog of tables.

```python
run_sql("""
    SELECT table_name
    FROM information_schema.tables
    WHERE table_schema = 'public'
    ORDER BY table_name
""", project)
```

The tables are empty for now, we'll add data in the next notebook. You can
peek at any table's structure with a query like this:

```python
run_sql("""
    SELECT column_name, data_type
    FROM information_schema.columns
    WHERE table_name = 'book'
    ORDER BY ordinal_position
""", project)
```
