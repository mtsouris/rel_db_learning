## The problem

In your project the `loan` table has columns `book_id` and `member_id`.
These fields are supposed to refer to an existing book and an existing member.

However, **nothing prevents** someone from inserting a loan with `book_id = 9999`,
even if no book with that `book_id` exists. The result would be a record
pointing to nothing.

## The solution: foreign key

A **foreign key** (Foreign Key, or **FK**) is a column in one table that
**references** the primary key of another table.

When we define an FK, the database guarantees that:
- **The value must exist** as a PK in the other table
- We cannot **delete** a record that is referenced elsewhere

## How it looks in pgModeler

In Chinook, the `album` table has an FK pointing to the `artist` table. Let
us look at both together:

```python
from IPython.display import SVG
SVG(filename="../../pgmodeler/exports/album_artist_with_fk_highlighted.svg")
```

## Reading the diagram

Notice the **line** connecting the two tables. This represents the FK
relationship.

In the `album` table you will see:

| Element | Meaning |
|---|---|
| `artist_id` marked `«fk»` | The column is a foreign key |
| `album_artist_id_fkey` | The name of the FK constraint |

The connecting line shows **which table the FK references**. In our case:
`album.artist_id` points to `artist_id`, which is the PK of `artist`.

## Important: FK points to a PK

A foreign key **always** references a column that is a primary key in the
other table. You cannot create an FK pointing to an arbitrary column — only
to a PK.

This is exactly why you needed to define the PKs first in the previous
notebook!

## Practice: link the tables in your project

In your project, the `loan` table must connect to:
- The `book` table (via the `book_id` column)
- The `member` table (via the `member_id` column)

### Step 1: Open your model

Open pgModeler and the file
`student_work/my_project/stage_1_initial_design.dbm`.

### Step 2: Add FK for the loan → book relationship

There are two ways to create an FK in pgModeler. We will use the simplest:
**via the Constraints tab of the table**.

1. **Double-click** the `loan` table to open the editing window
2. Go to the **Constraints** tab
3. Click the Add button
4. In the window that opens:
   - **Name**: `loan_book_fk`
   - **Constraint Type**: select `Foreign key`
   - In the **Columns** table, map: `book_id`
   - In the **Referenced column** table (from public.book): `book_id`
5. Click **Apply** to save the FK
6. Click **Apply** again to save the table

You will now see a **line** appear between the `loan` and `book` tables.
That line is the visual representation of the relationship!

### Step 3: Add FK for the loan → member relationship

Repeat the same process to link `loan` to `member`:

| Field | Value |
|---|---|
| Source column (loan) | `member_id` |
| Referenced table | `public.member` |
| Referenced column (member) | `member_id` |

### Step 4: Save

Click **File → Save** (Ctrl+S).

You should now have two lines starting from the `loan` table and ending at
the `book` and `member` tables.

### Try it yourself: what a foreign key protects

You have seen how a foreign key links a child table to its parent. Now see what that link actually protects you from.

In the Library, a loan is a child of two parents: the member who borrowed and the book that was borrowed. The widget below lets you try actions that would break those links, and watch the foreign key step in.

Try each one:

- Delete a member who still has loans. With RESTRICT, the deletion is blocked, because removing that member would leave their loans pointing at someone who no longer exists.
- Switch to CASCADE and delete the same member. Now the deletion is allowed, and the member's loans are removed along with them.
- Add a loan for a member or a book that does not exist. The foreign key does not allow it, because a loan must reference a real member and a real book.

A member or book with no loans, like Carla or Snow Crash, can always be deleted safely.

## Key takeaways

- **Foreign keys** link tables to each other
- They enforce **referential integrity** — references never point to nothing
- An FK **always** references a column that is a PK in the other table
- In pgModeler, the relationship appears as a **line** between tables
