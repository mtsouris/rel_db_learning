## Your SQL toolbox so far

- `SELECT col1, col2 FROM table` (use `*` for all columns)
- `WHERE condition` (operators: `=`, `!=`, `<`, `>`, `<=`, `>=`)
- `WHERE col LIKE 'pattern%'` (`%` matches any sequence of characters)
- **Combining conditions:** join them with `AND` or `OR`,
  e.g. `WHERE genre_id = 1 AND milliseconds > 300000`
- **Missing values:** test with `IS NULL` or `IS NOT NULL`,
  e.g. `WHERE composer IS NOT NULL`
- `ORDER BY col ASC | DESC` (ASC is the default)
- `SELECT DISTINCT col` (remove duplicates)
- Aggregates: `COUNT(*)`, `SUM(col)`, `AVG(col)`, `MIN(col)`, `MAX(col)`
- `GROUP BY col` (apply aggregate per group)
- `HAVING condition` (filter groups after aggregation)

A reminder on clause order:
`SELECT ... FROM ... WHERE ... GROUP BY ... HAVING ... ORDER BY ...`
