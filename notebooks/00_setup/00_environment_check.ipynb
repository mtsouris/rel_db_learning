### If the automatic installation failed

You can install Chinook manually via pgAdmin:

1. Open pgAdmin
2. Right-click `Databases` → `Create` → `Database...`
3. In the **Database** field type `chinook` and click **Save**
4. Right-click the new database `chinook` → **Query Tool**
5. In the Query Tool, click the **Open File** icon (folder icon, top-left)
6. Select the file: `src/rel_db_learning/data/chinook/chinook_postgres.sql`
7. Click **Execute** (▶ or F5)
8. In the pgAdmin Object Explorer, right-click `Databases` → **Refresh**
