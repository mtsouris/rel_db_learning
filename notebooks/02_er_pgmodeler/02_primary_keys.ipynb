## The problem

Imagine you have two members in the library with the same name. How would the
system tell apart who borrowed which book? With a name alone, we cannot. We need **something unique** for each member.

## The solution: primary key

A **primary key** (Primary Key, or **PK**) is a column (or combination of
columns) that:

- Has a **unique value** for every row in the table
- **Cannot be empty** (NULL)
- Does not change for the entire lifetime of the record

Every table in a relational database **must have** a primary key. It is the
«identifier code» of each row.

## Natural vs Synthetic keys

There are two philosophies for choosing a PK:

**Natural key**: a column that already exists and is unique in the real world.
For example, a company's **tax identification number**, or the **ISBN** of a
book.

**Synthetic key** (also called surrogate key): an artificial number we assign
ourselves, with no real-world meaning outside the database. It is typically
named `xxx_id` and increments automatically.

In your project we will use **synthetic keys** (`book_id`, `member_id`,
`loan_id`). This is the most common practice because:
- They do not depend on real-world data
- They never change
- They are small and efficient

## Practice: define PKs in your project

It is time to add primary keys to the tables you created in the previous
notebook.

### Step 1: Open your model

1. Open pgModeler
2. **File → Open** and select `student_work/my_project/stage_1_initial_design.dbm`

### Step 2: Define PK on the Book table

1. **Double-click** the `book` table to open the editing window
2. Go to the **Columns** tab
3. Find the **`book_id`** row
4. In the options at the bottom-left, check **Primary Key**
5. Click **Apply** again to save the table

You will now see the **«pk»** indicator next to the `book_id` column in the
diagram, and the constraint **`book_pk`** in the bottom section of the table.

### Step 3: Define PK on the other two tables

Repeat the same process:

| Table | Column to make PK |
|---|---|
| `member` | `member_id` |
| `loan` | `loan_id` |

### Step 4: Save

**File → Save** (Ctrl+S) to save the changes to
`stage_1_initial_design.dbm`.

## Key takeaways

- Every table **must have** a primary key
- The PK guarantees the **uniqueness** of each row
- In pgModeler it is shown with the **«pk»** indicator and the constraint
  named **`xxx_pk`**
- In our project we use **synthetic keys** (`xxx_id`)
