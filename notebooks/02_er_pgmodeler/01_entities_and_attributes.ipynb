## Entity: what is it?

In database modelling, an **entity** is any «thing» we want to store
information about.

Examples of entities from Chinook:
- **Artists** (artist)
- **Albums** (album)
- **Customers** (customer)

Each entity becomes a **table** in a relational database.

## Attribute: what is it?

Every entity has **characteristics** we want to record. These characteristics
are called **attributes** and become the **columns** of the table.

For example, the «artist» entity in Chinook has:
- A unique **identifier** (artist_id)
- A **name** (name)

## Reading the diagram

Each rectangle in pgModeler represents an **entity/table**. It has three
sections:

| Section | Contents |
|---|---|
| **Header (blue)** | The schema (`public.`) and the **name** of the table (`artist`) |
| **Columns** | The attributes of the entity, with their types |
| **Constraints** | Rules that must be satisfied (e.g. primary key) |

In the `artist` table you can see:

| Element | Meaning |
|---|---|
| `artist_id` | Column name — unique identifier for each artist |
| `integer` | Data type — an integer number |
| `name` | Column name — the artist's name |
| `varchar(120)` | Data type — text up to 120 characters |
| `«pk»` (green) | Indicates that this column is a **primary key** — you will learn about this shortly |
| `artist_pkey` | The name of the primary key constraint — generated automatically by pgModeler |

## Step 1: Open my_project in pgModeler

Instead of starting from an empty model, you will **import** the existing
(but empty) `my_project` database into pgModeler. This connects your diagram
to that specific database from the start — exactly as you would do in a real
professional environment.

### 1a. Open pgModeler

If it is not already open it.

### 1b. Start the Import

From the menu, click **File → Import** (or use the Import button in the side bar).

### 1c. Connect to the database

A wizard will open. Select the **connection** you have set up to your local
PostgreSQL instance.

If you have not yet set up a connection:

1. Click the **Edit Connections** icon (usually a gear or edit icon next to
   the dropdown)
2. Click **New** or **+**
3. Fill in:
   - **Alias**: a name of your choice, e.g. `Local PostgreSQL`
   - **Host**: `localhost`
   - **Port**: `5432`
   - **User name**: your username (typically `postgres`)
   - **Password**: your password
   - **Connect via**: leave the default
4. Click **Test** to verify the connection
5. If it succeeds, click **Save** or **Apply**

Select this connection from the dropdown.

### 1d. Select the database

1. In the list of available databases, find and select **`my_project`**.
2. Under **Schemas**, select **`public`**.
3. Options:
   - **Import system objects**: leave **OFF**
   - **Import extension objects**: leave **OFF**
   - **Auto-resolve dependencies**: leave **ON**
4. Click **Import**.

Because the database is **empty** (no tables yet), you will see a
**blank canvas**. This is expected.

### 1e. Initial save

Before adding anything, save the empty model as a file:

1. **File → Save As**

## Step 2: Create the Book table

In the **toolbar** on the left, find the **New → Schema Object → Table** tool
(or simply right-click on the canvas). An editing window will open.

In the **Name** field type: `book`

To add columns:
1. Go to the **Columns** tab
2. Click the first button (Add)
3. Fill in for each column: name and data type

Add the following columns to the `book` table:

| Name | Type | Note |
|---|---|---|
| `book_id` | integer | Unique identifier for each book |
| `title` | varchar(200) | The title |
| `authors` | text | The authors (there may be more than one) |
| `genres` | text | The genres / categories of the book |
| `publisher_name` | varchar(150) | The name of the publisher |
| `publisher_address` | varchar(200) | The publisher's address |
| `publisher_city` | varchar(100) | The publisher's city |

When you are done, click **Apply** to save the table.
