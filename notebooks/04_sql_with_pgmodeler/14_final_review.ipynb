## Showcase: queries that combine everything

Let's load the full Library dataset and run a few queries that bring together
joins, grouping, CTEs, and views.

```python
reset_my_project_data_full()
```

Who currently has a book out? The current_loans view answers instantly:

```python
run_sql("SELECT * FROM current_loans ORDER BY loan_id LIMIT 10", project)
```

The most active members, using a CTE to count loans then joining member names:

```python
run_sql("""
    WITH loan_counts AS (
        SELECT member_id, COUNT(*) AS cnt FROM loan GROUP BY member_id
    )
    SELECT m.full_name, lc.cnt AS loans
    FROM member m
    JOIN loan_counts lc ON m.member_id = lc.member_id
    ORDER BY lc.cnt DESC, m.full_name
    LIMIT 5
""", project)
```

And the overdue loans, straight from the view:

```python
run_sql("SELECT * FROM overdue_loans ORDER BY loan_id", project)
```

## Bonus: a last look at Ibis

Throughout Module 4 you saw Ibis as a Python way to express queries that
compile to SQL. As a final reminder here is an aggregate in Ibis and the SQL
it generates.

```python
import ibis
from rel_db_learning import get_ibis_chinook

ibis_chinook = get_ibis_chinook()
track = ibis_chinook.table("track")

result = track.group_by("genre_id").aggregate(track_count=track.count())
print(ibis.to_sql(result))
```

Ibis is a bridge to data analytics workflows in Python. The SQL you learned
here is what tools like Ibis generate under the hood.
