# What Is a Relation?

In the previous notebook we learned what a database is. Now we will see how data is
organised within it.

In a relational database, the central concept is the **relation**, which in practice
is a **table**.

## Example

Let us look again at the `artist` table from Chinook:

```python
from rel_db_learning import get_chinook_connection, run_sql

chinook = get_chinook_connection()
run_sql("SELECT * FROM artist LIMIT 5", chinook)
```

Notice the structure:

- The table is called **`artist`** — this is the name of the relation.
- It has two **columns**: `artistid` and `name`. Each column is called an **attribute**
  and has a specific data type (number, text, etc.).
- Each **row** of the table is called a **tuple** and represents a specific artist.

## Terminology

In relational databases we use the following terms:

| Everyday term | Formal term |
|---|---|
| Table | Relation |
| Column | Attribute |
| Row | Tuple |
| Kind of data | Type (Domain/Data type) |

We will use both terms the same way. When we say "table Album" we mean exactly
the same as "relation Album".

## Schema vs Instance

Two important concepts:

- The **schema** is the **structure** of the table: which columns it has, their types,
  and which rules apply. It rarely changes.
- The **instance** is the **data** that exists in the table at a particular moment.
  It changes continuously (new records, deletions, updates).

For example, the `artist` table has the same schema (columns artistid, name) regardless
of whether it contains 10 or 10,000 artists.

## Multiple tables, one database

A database does not have just one table, it has many and they are related to each other.
Let us look at another Chinook table, `album`:

```python
run_sql("SELECT * FROM album LIMIT 10", chinook)
```

The `album` table contains albums. Each album has:

- Its own **identifier** (`album_id`)
- A **title** (`title`)
- A reference to the **artist** who released it (`artist_id`)

The `artist_id` column in `album` refers to the `artist_id` column in `artist`. This is
how the two tables are **linked**. In latethe next modules we will look in detail at how these
links work.

## How many tables are there?

Chinook consists of 11 tables in total. Let us list them:

```python
run_sql("""
    SELECT table_name 
    FROM information_schema.tables 
    WHERE table_schema = 'public'
    ORDER BY table_name
""", chinook)
```

These are all the tables in the database. In the next notebook we will explore them in
detail to understand what each one contains.

## Key takeaways

- A **relation** is a **table** with columns (attributes) and rows (tuples).
- The **schema** is the structure; the **instance** is the data.
- A database consists of **many tables** that may be linked to each other.
