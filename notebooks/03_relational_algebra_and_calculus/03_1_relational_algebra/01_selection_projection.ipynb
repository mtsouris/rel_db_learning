# Selection and Projection

In this notebook we will explore **relational algebra** a way to describe
queries over tables.

Below you will find a table of the symbols we will use.
You can copy symbols from here when writing your own expressions.

| Symbol | Name | What it does |
|---|---|---|
| σ | Selection | Selects rows that satisfy a condition |
| π | Projection | Selects columns |
| ⋈ | Join | Combines two tables |
| ∪ | Union | Union of two sets |
| ∩ | Intersection | Intersection of two sets |
| − | Difference | Difference of two sets |
| ∧ ∨ ¬ | AND OR NOT | Logical operators |

In this notebook we will cover the first two: **σ** and **π**.

## What is relational algebra

Relational algebra is a procedural language: it describes step by step
how to reach a result by applying operations to tables.

Each operation takes one or more tables and produces a **new table** as its
result. This means we can chain operations: the output of one becomes the
input of the next.

## Selection (σ): filtering rows

**Selection** (symbol **σ**, the Greek letter «sigma») selects the **rows**
of a table that satisfy a condition.

Syntax: **σ(condition)(table)**

For example, to find the album with `album_id = 1`:

We use `=` to select rows with a specific value.
Example: the album with `album_id = 1`.

```python
from rel_db_learning.algebra.runner import student_algebra
from rel_db_learning.widgets.self_assessment import multiple_choice, algebra_check

student_algebra("σ(album_id = 1)(album)")
```

We can also use comparison operators: `=`, `!=`, `<`, `>`, `<=`, `>=`.

Example: albums by artists with `artist_id` less than 5.

```python
student_algebra("σ(artist_id < 5)(album)")
```

The **LIKE** operator allows pattern-based text matching.
`%` means «any characters».

Example: artists whose name begins with «A».

```python
student_algebra("σ(name LIKE 'A%')(artist)")
```

We can combine conditions with logical operators **∧** (AND), **∨** (OR), **¬** (NOT).

Example: albums with `artist_id` greater than 5 **and** less than 10.

```python
student_algebra("σ(artist_id > 5 ∧ artist_id < 10)(album)")
```

## Projection (π): selecting columns

**Projection** (symbol **π**, the Greek letter «pi») selects specific
**columns** of a table, ignoring the rest.

Syntax: **π(column1, column2, ...)(table)**

To see only the artist names (not the artist_id):

To see only the artist names, we select a single column.

```python
student_algebra("π(name)(artist)")
```

We can select multiple columns, separated by commas.

```python
student_algebra("π(album_id, title)(album)")
```

Projection works on any table.
Example: only the titles from the `album` table.

```python
student_algebra("π(title)(album)")
```

```python
multiple_choice(
    question="What does projection (π) do?",
    options=[
        "Selects rows based on a condition",
        "Selects specific columns",
        "Combines two tables",
    ],
    correct_index=1,
    explanation="Projection (π) keeps only the columns we request.",
)
```

```python
multiple_choice(
    question="Which expression returns only the album titles?",
    options=[
        "σ(title)(album)",
        "π(title)(album)",
        "π(album)(title)",
    ],
    correct_index=1,
    explanation="π(title)(album) keeps only the title column from the album table.",
)
```

## Combining σ and π

The power of algebra lies in **combining** operations. The output of one
becomes the input of the next.

For example: «I want the **titles** of albums by the artist with
`artist_id = 22`».

This requires two steps:
1. **Selection**: keep the rows where `artist_id = 22`
2. **Projection**: keep only the `title` column

We write this «from the inside out»: selection first, then projection
wrapped around it.

```python
student_algebra("π(title)(σ(artist_id = 22)(album))")
```

Notice the structure: **σ** is applied first (to the `album` table), then
**π** is applied to the result. Read it from the inside out,
as in mathematics: f(g(x)).

```python
multiple_choice(
    question="What does the expression  π(name)(σ(artist_id < 10)(artist))  do?",
    options=[
        "Returns all columns of artists with artist_id < 10",
        "Returns only the names of artists with artist_id < 10",
        "Returns the names of all artists",
    ],
    correct_index=1,
    explanation="σ filters (artist_id < 10), π keeps only the name column.",
)
```

```python
multiple_choice(
    question="In what order do we read the expression  π(title)(σ(artist_id = 5)(album)) ?",
    options=[
        "π first, then σ",
        "σ first (inner), then π (outer)",
        "The order does not matter",
    ],
    correct_index=1,
    explanation="We read from the inside out: σ is applied to album first, then π is applied to the result.",
)
```

## What we learned

- **Relational algebra** is a procedural query language
- **Selection (σ)** selects **rows** based on a condition
- **Projection (π)** selects **columns**
- We can **combine** operations: the output of one becomes the input of the next
- We read composite expressions **from the inside out**
