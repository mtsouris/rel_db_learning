## Why does the type matter?

A data type defines **what can be stored** in a column. With the right type:

- **You ensure integrity**: no one can store «apple» in an age column
- **You save space**: an integer takes up less space than a string
- **You enable correct operations**: you can add numbers, compare dates,
  and sort correctly
- **You avoid errors**: the database will reject records with the wrong type
  of value

## Core PostgreSQL data types

PostgreSQL has many available types. At the level of this course, the
following cover almost every situation:

### Numeric types

| Type | Description | Example value |
|---|---|---|
| `integer` (or `int`) | Integer number | `42`, `-7` |
| `bigint` | Large integer (e.g. for large IDs) | `9999999999` |
| `numeric(p, s)` | Decimal number with precision | `numeric(10,2)` → `123.45` |
| `real` / `double precision` | Floating-point decimal number | `3.14159` |

### Text types

| Type | Description | Example value |
|---|---|---|
| `varchar(n)` | Text up to `n` characters | `varchar(100)` → `"Hello"` |
| `text` | Text with no length limit | `"Long content..."` |
| `char(n)` | Fixed-length text of `n` characters (padded with spaces) | `char(3)` → `"USA"` |

### Date & time types

| Type | Description | Example value |
|---|---|---|
| `date` | Date only | `'2025-05-15'` |
| `time` | Time only | `'14:30:00'` |
| `timestamp` | Date and time | `'2025-05-15 14:30:00'` |
| `timestamptz` | Date/time with time zone | `'2025-05-15 14:30+02'` |

### Boolean type

| Type | Description | Values |
|---|---|---|
| `boolean` (or `bool`) | True or false | `true`, `false`, `null` |

## varchar vs text: which one to choose?

Two text types that can be confusing:

**`varchar(n)`** — when you have a **limit** on the length (e.g. a name up to
150 characters). The database enforces this limit.

**`text`** — when you **do not know** the length in advance, or it is
potentially large (e.g. a description, a comment).

### General rule

Choose `varchar(n)` when:
- You have a reasonable choice for `n` (e.g. a city name up to 100 characters)
- You want the database to enforce that limit

Choose `text` when:
- The size can vary greatly
- You do not want a limit
