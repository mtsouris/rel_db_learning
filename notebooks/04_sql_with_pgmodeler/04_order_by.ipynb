# Order By Results

So far your queries returned rows in whatever order the database chose, and
all matching rows at once. In this notebook you will control the order of
results with **ORDER BY** and remove duplicates with **DISTINCT**.

## ORDER BY: sorting results

ORDER BY sorts the rows by one or more columns:

```sql
SELECT columns FROM table ORDER BY column [ASC | DESC];
```

ASC is ascending (the default), DESC is descending. Let's sort Chinook
artists alphabetically by name:

````python
from rel_db_learning import get_chinook_connection, get_project_connection, run_sql
from rel_db_learning.widgets.self_assessment import sql_query_check

chinook = get_chinook_connection()
project = get_project_connection()

run_sql("SELECT name FROM artist ORDER BY name", chinook)
````

To sort in reverse (Z to A), use DESC:

````python
run_sql("SELECT name FROM artist ORDER BY name DESC", chinook)
````

## LIMIT: returning only some rows

LIMIT restricts how many rows come back. This is useful for "top N" style
queries, especially combined with ORDER BY. The first 5 artists
alphabetically:

````python
run_sql("SELECT name FROM artist ORDER BY name LIMIT 5", chinook)
````

Combine ORDER BY DESC with LIMIT to get the "last" or "highest" rows. The 5
albums with the highest album_id:

````python
run_sql("SELECT album_id, title FROM album ORDER BY album_id DESC LIMIT 5", chinook)
````

## DISTINCT: removing duplicates

DISTINCT keeps only unique values, removing repeats. For example, many
tracks share the same composer. To see each composer once:

````python
run_sql("SELECT DISTINCT composer FROM track ORDER BY composer LIMIT 20", chinook)
````

## Now to your Library

Let's apply these to your own data. Your books, sorted alphabetically:

````python
run_sql("SELECT title FROM book ORDER BY title", project)
````

Your books come from only a few publisher cities. DISTINCT shows which ones:

````python
run_sql("SELECT DISTINCT publisher_city FROM book ORDER BY publisher_city", project)
````

## The same in Python with Ibis

In the previous notebook you met Ibis. Here is how ordering and limiting
look in Ibis, on the Chinook database.

````python
import ibis
from rel_db_learning import get_ibis_chinook

ibis_chinook = get_ibis_chinook()
artist = ibis_chinook.table("artist")

# First 5 artists alphabetically
result = artist.order_by("name").select("name").limit(5)
result.execute()
````

And the SQL Ibis generated for that query:

````python
print(ibis.to_sql(result))
````
