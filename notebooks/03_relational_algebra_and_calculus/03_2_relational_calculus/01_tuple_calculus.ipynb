# Module 03_2: Relational Calculus

In the previous part we studied **relational algebra**, which describes
**how** (in what steps) to retrieve data. Now we will look at a different
approach: **relational calculus**.

The key difference:
- **Algebra** (procedural): «Take the artist table, filter with σ, keep
  columns with π» you describe the **steps**.
- **Calculus** (declarative): «I want the names of artists with artist_id < 10»
 you describe **what** you want, not how to get it.

Both languages are **equivalent** they express the same queries in
different ways.

Symbol palette

{ }     |     ∈     .     ∧     ∨     ¬     =     <     >     LIKE

The general form of a tuple calculus expression is:

```
{ t | t ∈ R ∧ condition }
```

Reads as: «**the set of tuples t** such that **t belongs to relation R**
**and** satisfies the **condition**».

- `t` is a **tuple variable** it represents a row.
- `t ∈ R` says that the row comes from table R.
- The condition after `∧` filters which rows we keep.

## Selection: selecting rows

To select rows, we add a condition after `∧`.

Example: all artists with `artist_id` less than 5.

````python
from rel_db_learning.algebra.runner import student_calculus
from rel_db_learning.parsers.algebra_to_calculus import algebra_to_calculus
from rel_db_learning.widgets.self_assessment import multiple_choice

student_calculus("{ t | t ∈ artist ∧ t.artist_id < 5 }")
````

Second example: albums of the artist with `artist_id = 90`.

````python
student_calculus("{ t | t ∈ album ∧ t.artist_id = 90 }")
````

Third example: artists whose name starts with «B»
(using LIKE with %).

````python
student_calculus("{ t | t ∈ artist ∧ t.name LIKE 'B%' }")
````

## Projection: selecting columns

To keep only specific columns, we declare them **at the beginning**,
before the `|`, in the form `t.column`.

Example: only the names of artists.

````python
student_calculus("{ t.name | t ∈ artist }")
````

We can combine projection with selection. Only the names of artists
with `artist_id` less than 5:

````python
student_calculus("{ t.name | t ∈ artist ∧ t.artist_id < 5 }")
````

Multiple columns, separated by a comma:

````python
student_calculus("{ t.album_id, t.title | t ∈ album ∧ t.artist_id = 90 }")
````

## Correspondence with algebra

Since the two languages are equivalent, every algebra expression has a
corresponding calculus expression. Let's see it in practice: we'll take
algebra expressions and see the equivalent calculus form.

````python
print("Algebra:  σ(artist_id < 10)(artist)")
print("Calculus:", algebra_to_calculus("σ(artist_id < 10)(artist)"))
print()
print("Algebra:  π(name)(artist)")
print("Calculus:", algebra_to_calculus("π(name)(artist)"))
print()
print("Algebra:  π(name)(σ(artist_id < 10)(artist))")
print("Calculus:", algebra_to_calculus("π(name)(σ(artist_id < 10)(artist))"))
````

Notice the correspondence:
- **σ(condition)** in algebra becomes **∧ condition** in calculus
- **π(columns)** in algebra becomes **t.columns** at the beginning of the calculus expression

It is the same way of thinking, expressed differently.

## What wEsaw in this notebook

- **Relational calculus** is **declarative**: it describes **what** we want
- The syntax: **{ t.columns | t ∈ R ∧ condition }**
- **Columns** (projection) are declared before `|`
- The **condition** (selection) goes after `∧`
- Calculus and algebra are **equivalent** — they describe the same queries
