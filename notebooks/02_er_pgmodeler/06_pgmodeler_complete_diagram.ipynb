## How to read the results

The report is divided into four sections:

|---|---|

| **Columns** | Whether each table has the required columns |
| **Primary Keys** | Whether the correct columns are defined as PRIMARY KEY |

Each line has an icon:

- `✓` — the check passed
- `✗` — the check failed, correction needed
- `i` — informational note (e.g. different data type), does not affect the result

## If you saw failures

**Table not found**
Make sure the table name is spelled exactly as required (e.g. `book`, not
`books` or `Book`). pgModeler stores names exactly as you type them.

**Column not found**
Check that the column exists in the correct table and that its name is exact.
Remember that FK columns created automatically by the Relationships tool do
not need to be added manually.

**Primary key missing**
Open the table in pgModeler, go to the **Constraints** tab, and make sure
there is a TYPE PRIMARY KEY that includes the correct column.

**Relationship missing or wrong cardinality**
You can define relationships in two ways:
- **Foreign Keys** tab on the table: add the FK column and set the constraint.
- **Relationships** tool in the toolbar: draw the line from one table to the
  other.

For 1:1, the FK column must also have a UNIQUE constraint. If you are using
the Relationships tool, select **One to One** from the relationship options.

### Answers

1. **11 tables**: artist, album, track, genre, media_type, customer, invoice,
   invoice_line, employee, playlist, playlist_track
2. **`track`** is the most «central» table — it connects to `album`, `genre`,

3. **`playlist_track`**: it is the junction table linking `track` and

4. **1:N**: one customer (`customer`) can have many invoices (`invoice`), but
