# Set Operations: ∪, ∩, −

In this notebook we will look at the three **set operations** of
relational algebra: union, intersection, and difference.

Symbol reminder (copy whichever you need):

| Symbol | Name | What it does |
|---|---|---|
| σ | Selection | Selects rows that satisfy a condition |
| π | Projection | Selects columns |
| ⋈ | Join | Combines two tables |
| ∪ | Union | Union of two sets |
| ∩ | Intersection | Intersection of two sets |
| − | Difference | Difference of two sets |
| ∧ ∨ ¬ | AND OR NOT | Logical operators |

## Important rule: compatible schemas

Set operations only work between tables with **the same columns**
(same count, same types). This is called **union compatibility**.

For example, we **cannot** do `artist ∪ album` because they have different
columns.

However, we **can**:
- Apply set operations to **the same table** with different conditions
- Use **projection** to give the two sets the same columns

## Union (∪)

**Union** returns all rows that appear in the first **or** the second set
(without duplicates).

Example: artists whose name starts with «A» **or** «B».

```python
from rel_db_learning.algebra.runner import student_algebra
from rel_db_learning.widgets.self_assessment import multiple_choice, algebra_check

student_algebra("σ(name LIKE 'A%')(artist) ∪ σ(name LIKE 'B%')(artist)")
```

Notice that both sides of the union operate on **the same table**
(`artist`), so they automatically have the same columns — they are compatible.

We can union two portions of **the same table** using numeric conditions.
Example: albums at the beginning **or** at the end of the catalogue.

```python
student_algebra("σ(album_id < 5)(album) ∪ σ(album_id > 340)(album)")
```

## Using projection for compatibility

When we want a set operation between **different** tables, we use
projection to keep only the common columns.

For example, we want all `artist_id` values that appear
either in the `artist` table or in the `album` table. Both tables have
an `artist_id` column, so projection makes them compatible:

```python
student_algebra("π(artist_id)(artist) ∪ π(artist_id)(album)")
```

## Intersection (∩)

**Intersection** returns only the rows that appear **in both** sets.

Example: artists whose name starts with «A» **and** also contains
«c» somewhere in their name.

```python
student_algebra("σ(name LIKE 'A%')(artist) ∩ σ(name LIKE '%c%')(artist)")
```

The result contains only the artists that satisfy **both** conditions
simultaneously.

Intersection also works with numeric conditions.
Example: albums that satisfy **both** conditions simultaneously.

```python
student_algebra("σ(album_id < 20)(album) ∩ σ(artist_id < 5)(album)")
```

Another example: artists whose name starts with «M» **and** contains the letter «e».

```python
student_algebra("σ(name LIKE 'M%')(artist) ∩ σ(name LIKE '%e%')(artist)")
```

## Difference (−)

**Difference** returns the rows of the first set that do **not** appear
in the second.

Example: artists whose name starts with «A» but does **not** contain «c».

```python
student_algebra("σ(name LIKE 'A%')(artist) − σ(name LIKE '%c%')(artist)")
```

Note the difference symbol: it is **−** (minus), which you can copy
from the palette. The result is «those starting with A» **minus** «those
containing c».

Example with numeric conditions: the first albums in the catalogue excluding
those by the first artists.

```python
student_algebra("σ(album_id < 20)(album) − σ(artist_id < 5)(album)")
```

Another example: artists whose name starts with «M» but does **not** contain the letter «e».

```python
student_algebra("σ(name LIKE 'M%')(artist) − σ(name LIKE '%e%')(artist)")
```

```python
multiple_choice(
    question="What does difference (−) do?",
    options=[
        "Returns the rows common to both sets",
        "Returns the rows of the first set that do NOT appear in the second",
        "Combines two sets",
    ],
    correct_index=1,
    explanation="Difference A − B keeps what is in A but not in B.",
)
```

```python
multiple_choice(
    question="Does the order matter in difference? That is, is  A − B  the same as  B − A ?",
    options=[
        "Yes, they are always the same",
        "No, the order matters (A − B ≠ B − A in general)",
        "Difference does not accept two sets",
    ],
    correct_index=1,
    explanation="Difference is not commutative. A − B keeps elements of A, while B − A keeps elements of B.",
)
```

## What we learned 

- **Union (∪)** combines rows from two sets (or)
- **Intersection (∩)** keeps only the rows common to both sets (and)
- **Difference (−)** subtracts from the first set whatever appears in the second
- Set operations require **compatible schemas** (same columns)
- We use **projection** to make two tables compatible
