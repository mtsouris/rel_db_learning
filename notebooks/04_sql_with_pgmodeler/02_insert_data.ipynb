# Inserting Data with INSERT

Your Library database has tables, but they are empty. In this notebook you
will learn the **INSERT** statement, which adds rows to a table. You will
write a few inserts by hand to learn the syntax, then load a full dataset
so the rest of the module has plenty of data to work with.

## The INSERT statement

The basic form is:

```sql
INSERT INTO table_name (column1, column2, ...)
VALUES (value1, value2, ...);
```

You list the columns you are filling, then provide a matching list of
values. Text values go in single quotes, numbers do not.

Because INSERT does not return any rows (unlike SELECT), we run it with
`run_ddl` rather than `run_sql`.

````python
from rel_db_learning import get_project_connection, run_sql, run_ddl, reset_my_project_data

project = get_project_connection()
````

## Start clean

Let's start from an empty set of tables so your hand-written inserts are
easy to see. The following removes any existing rows (the schema stays
intact).

````python
run_ddl("TRUNCATE loan, member_card, book, member RESTART IDENTITY CASCADE;", project)
````

## Insert your first book

Let's add a single book. Notice that the `authors` and `genres` columns hold
plain text. We will store multiple authors and genres as comma-separated
text for now.

````python
run_ddl("""
    INSERT INTO book (book_id, title, authors, genres, publisher_name, publisher_address, publisher_city)
    VALUES (1, 'The Hobbit', 'J.R.R. Tolkien', 'Fantasy, Adventure', 'George Allen & Unwin', '40 Museum Street', 'London');
""", project)
````

Let's confirm it was added:

````python
run_sql("SELECT * FROM book", project)
````

## Insert several rows at once

You can insert multiple rows in one statement by separating the value lists
with commas:

````python
run_ddl("""
    INSERT INTO book (book_id, title, authors, genres, publisher_name, publisher_address, publisher_city)
    VALUES
        (2, 'The Fellowship of the Ring', 'J.R.R. Tolkien', 'Fantasy, Adventure', 'George Allen & Unwin', '40 Museum Street', 'London'),
        (3, 'Good Omens', 'Terry Pratchett, Neil Gaiman', 'Fantasy, Comedy', 'Gollancz', '14 Henrietta Street', 'London');
""", project)

run_sql("SELECT * FROM book", project)
````

## A closer look at the data

Look carefully at the book rows you inserted. A few things are worth noticing:

- The `authors` column sometimes holds more than one name in a single text

- The `genres` column also packs several genres into one text value.
- The publisher information (name, address, city) is repeated in full on

These choices keep the design simple for now, but they will cause problems
