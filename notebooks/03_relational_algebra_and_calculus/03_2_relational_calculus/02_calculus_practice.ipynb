## Symbol palette
`{ }    |    ∈    .    ∧    ∨    ¬    =    !=    <    >    <=    >=    LIKE`

Syntax reminder:
- **{ t | t ∈ R ∧ condition }** selects rows
- **{ t.column | t ∈ R ∧ condition }** selects columns
- Columns are declared **before** the `|`, the condition goes **after** `∧`
