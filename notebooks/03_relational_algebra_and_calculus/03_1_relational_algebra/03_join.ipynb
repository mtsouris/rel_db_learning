# Join (⋈): Combining Tables

So far we have worked with one table at a time. But the real power of
relational databases lies in **combining** data from multiple tables. This is
what the **join (⋈)** operation does.

## Symbol palette

| Symbol | Name | What it does |
|---|---|---|
| σ | Selection | Selects rows that satisfy a condition |
| π | Projection | Selects columns |
| ⋈ | Join | Combines two tables |
| ∪ | Union | Union of two sets |
| ∩ | Intersection | Intersection of two sets |
| − | Difference | Difference of two sets |
| ∧ ∨ ¬ | AND OR NOT | Logical operators |

## The problem

In Chinook, the `album` table has only `album_id`, `title`, and `artist_id`.
It does not contain the **name** of the artist — only the `artist_id`.

If we want to see «album title **together with** artist name», we need to
**combine** the two tables. This is where join comes in.

## Natural Join (⋈)

The **natural join** (symbol **⋈**) combines two tables by matching their
rows on the basis of **common columns** (columns with the same name).

Syntax: **table1 ⋈ table2**

The `artist` and `album` tables share the column `artist_id`. The natural
join matches them on that column:

```python
from rel_db_learning.algebra.runner import student_algebra
from rel_db_learning.widgets.self_assessment import multiple_choice, algebra_check

student_algebra("artist ⋈ album")
```

Notice that each row in the result combines information from both tables:
`name` (from artist) alongside `title` (from album). The matching was done
automatically via the shared `artist_id` column.

## Second example: album with track

The `album` and `track` tables share the column `album_id`. Let us look at
the tracks of the album with `album_id = 1`:

```python
student_algebra("σ(album_id = 1)(album ⋈ track)")
```

Here we combined join with selection: first we joined `album` and `track`,
then kept only the rows for album 1.

## Third example: the natural join pitfall

The natural join matches on **all** common columns. This can sometimes cause
problems.

Try joining the `track` and `genre` tables:

```python
student_algebra("track ⋈ genre")
```

**There are no results!**

The reason: the `track` and `genre` tables have **two** common columns:
`genre_id` **and** `name` (a track has a `name`, and a genre has a `name`
too). The natural join tried to match on both meaning it looked for tracks
and genres with the **same name**, which does not exist.

**Therefore**: Natural join is convenient, but you must pay attention to the
common columns. When two tables share columns that you did not intend to join
on (like `name`), the result can be wrong or empty.

## Combining join with σ and π

Like all operations, join combines with selection and projection.

Example: «I want the **names** of artists together with the **titles** of their
albums, but only for the artist with `artist_id = 1`»:

```python
student_algebra("π(name, title)(σ(artist_id = 1)(artist ⋈ album))")
```

Read from the inside out:
1. **artist ⋈ album**: join the two tables
2. **σ(artist_id = 1)**: keep only artist 1
3. **π(name, title)**: keep only the name and title columns

## What we saw

- **Natural join (⋈)** combines two tables based on their **common columns**
- It allows us to combine data that lives in different tables
- **Caution**: natural join matches on **all** common columns if there are
  columns you did not intend to join on (like `name`), the result can be empty
  or wrong
- Join combines with σ and π for more complex or specific queries
