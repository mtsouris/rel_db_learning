## So what is a Database?

A **database** is an organised system for storing information, designed for:

- **Storing** large volumes of data
- **Retrieving** specific information quickly
- **Updating** data safely
- **Concurrent use** by many users

In this module we will work with **relational databases**, which organise data in tables
like a spreadsheet, but far more powerful and capable.

## Why is Excel not enough?

Excel is excellent for small, personal use. However, as data grows, limitations appear:

| Problem in Excel | Solution in a DB |
|---|---|
| Hard to manage once data reaches thousands of rows | Handles millions of rows with ease |
| No integrity checks (e.g. you can type "apple" in an age column) | Enforces rules (types, constraints, relationships) |
| Hard to work on simultaneously with others | Designed for concurrent access |
| Data duplication (e.g. the same customer name in 100 rows) | Avoids duplication through table relationships |
| Complex queries are difficult | Powerful query language (SQL) |
