# A Tour of Chinook

In the previous notebook we saw that Chinook consists of 11 tables. In this notebook
we will explore them so you can understand what each one contains and how they are related.

Chinook models a digital music store. Its tables are divided into **three groups**:

1. **Music**: what we sell
2. **Sales**: who buys what
3. **Organization**: who works at the store and how it works

## Chinook Schema

Before looking at the tables one by one, let us look at the **schema** of Chinook.
Each rectangle below is a table, and the lines connecting them show the **relationships**
between them.

You do not need to understand all the details now — many symbols and elements are still
unfamiliar. The goal is to form a **general picture** before we dive into the specifics.

## Group 1: Music

The largest group. Tables:

- **`artist`** — the artists
- **`album`** — the albums (each album belongs to one artist)
- **`track`** — the tracks (each track belongs to one album)
- **`genre`** — the music genres (rock, jazz, classical, etc.)
- **`mediatype`** — the file formats (MPEG, AAC, etc.)

```python
from rel_db_learning import get_chinook_connection, run_sql

chinook = get_chinook_connection()

# How many artists, albums, and tracks do we have?
run_sql("""
    SELECT 
        (SELECT COUNT(*) FROM artist) AS kallitexnes,
        (SELECT COUNT(*) FROM album) AS diskoi,
        (SELECT COUNT(*) FROM track) AS tragoudia
""", chinook)
```

```python
# Sample tracks
run_sql("SELECT track_id, name, milliseconds, unit_price FROM track LIMIT 15", chinook)
```

Notice that each track has a duration in milliseconds and a price. `track` is the
most data-rich table in the group.

```python
# Available music genres
run_sql("SELECT * FROM genre", chinook)
```

## Group 2: Sales

Tables:

- **`customer`** — the customers
- **`invoice`** — the invoices (each invoice belongs to one customer)
- **`invoice_line`** — the line items of each invoice (which tracks were purchased)

```python
# Sample customers
run_sql("SELECT customer_id, first_name, last_name, country FROM customer LIMIT 10", chinook)
```

```python
# Which countries are our customers from?
run_sql("SELECT DISTINCT country FROM customer ORDER BY country", chinook)
```

```python
# Sample invoices
run_sql("SELECT invoice_id, customer_id, invoice_date, total FROM invoice LIMIT 10", chinook)
```

Each invoice has a `customer_id` that tells us which customer it belongs to. This is a
**relationship** between tables — we will look in detail at how it works in the next module.

## Group 3: Organization

Tables:

- **`employee`** — the store's employees
- **`playlist`** — the built-in playlists
- **`playlisttrack`** — which tracks belong to which playlists

```python
# Employees
run_sql("SELECT employee_id, first_name, last_name, title FROM employee", chinook)
```

```python
# Playlists
run_sql("SELECT * FROM playlist", chinook)
```

Notice the `playlist_track` table: it is a separate table that links playlists to tracks.
This is because a playlist can contain many tracks **and** a track can appear in many
playlists. Such relationships are called **many-to-many** and we will examine them in
detail soon.

```python
# How many tracks does each playlist have?
run_sql("""
    SELECT playlist_id, COUNT(*) AS arithmos_tragoudiwn 
    FROM playlist_track 
    GROUP BY playlist_id 
    ORDER BY playlist_id
""", chinook)
```

## First look at pgModeler

The diagram of the **Chinook** schema we saw above is static. **pgModeler** is the tool
we used to produce it, and it is the same tool you will use from the next module onwards
to design databases.

Let us open it now to get familiar with the environment.

### Step 1: Open pgModeler

If you have not run it before, find it in your programs menu and open it.

### Step 2: Open the Chinook file

1. From the menu click **File → Open Model**
2. Navigate to your project folder
3. Open the `pgmodeler/` folder
4. Select the file **`chinook.dbm`**
5. Click **Open**

Alternatively, you can import the database:

1. From the menu click **File → import** (or use the toolbar on the left)
2. Connection: select your local PostgreSQL connection. If none exists: click the
   Edit Connections icon, click New or +, and enter: Alias (e.g. "Local"),
   Host (localhost), Port (5432), Database (postgres), User, Password; then test
   the connection and click Save.
3. Database: select chinook from the dropdown
4. Schemas: select public; in the options check "Random color for relationships"
   and "Automatically resolve dependences"
5. Click **Import**

You will see the same diagram as above, but now it is **interactive**.

### Step 3: Navigate the canvas

Try the following to get comfortable:

- **Zoom in/out**: Hold `Ctrl` and scroll the mouse wheel. Alternatively, use the
  zoom buttons at the bottom of the window.

- **Pan the canvas**: Hold the middle mouse button (scroll wheel) and drag.
  Alternatively, use the scroll bars.

- **Move a table**: **Click and drag** a table to reposition it. Try arranging them
  however you prefer.

- **Select multiple objects**: Hold `Ctrl` and click multiple items, or drag a
  selection box around them.

### Step 4: Explore a table

1. Find the **`artist`** table in the diagram.
2. **Double-click** it.
3. An editing window will open with all the details:
   - **Columns**: the table's columns and their types
   - **Constraints**: the rules (primary keys, foreign keys, etc.)
   - **Triggers**, **Rules**, etc. (advanced — ignore these for now)
4. Click **Cancel** to close the window without making changes.

Try the same with other tables — for example `album` or `track`.

### Step 5: Explore a relationship

1. Find the **line** that connects the `album` and `artist` tables.
2. **Double-click** the line.
3. A window will open with details about the relationship: which columns create it,
   what type of association it is, etc.
4. Click **Cancel**.
