## SELECT: choosing what to see

The SELECT statement reads rows from a table. It lists the columns you want, followed by the table:

```sql
SELECT column1, column2 FROM table_name;
```

To see every column, use `*`. Let's look at artists in Chinook:

````python
from rel_db_learning import get_chinook_connection, get_project_connection, run_sql
from rel_db_learning.widgets.self_assessment import sql_query_check

chinook = get_chinook_connection()
project = get_project_connection()

run_sql("SELECT * FROM artist", chinook)
````

We can choose specific columns instead of all of them:

````python
run_sql("SELECT name FROM artist", chinook)
````

## WHERE: filtering rows

Most of the time you want only some rows, not all. The WHERE clause filters
rows by a condition:

```sql
SELECT columns FROM table WHERE condition;
```

Comparison operators: `=`, `!=` (or `<>`), `<`, `>`, `<=`, `>=`.

Let's find the album with album_id 1:

````python
run_sql("SELECT * FROM album WHERE album_id = 1", chinook)
````

Text values go in single quotes. Let's find the artist named 'Metallica':

````python
run_sql("SELECT * FROM artist WHERE name = 'Metallica'", chinook)
````

## LIKE: pattern matching

To search text by pattern, use LIKE with the `%`, which matches any
sequence of characters. This finds artists whose name starts with 'A':

````python
run_sql("SELECT name FROM artist WHERE name LIKE 'A%'", chinook)
````

You can place `%` anywhere. To find names that contain 'ero' anywhere:

````python
run_sql("SELECT name FROM artist WHERE name LIKE '%ero%'", chinook)
````

## Now your Library

## The same in Python with Ibis

Now for something different. **Ibis** is a Python library that lets you write queries in Python, which it then translates into SQL. If you go on to work in data analytics, you will meet tools like this. You do not need to memorize any of it, this section is just a preview.

The cell below installs Ibis if you do not already have it. It is a one-time setup, and it is optional.

````python
try:
    import ibis
    print("Ibis is ready.")
except ImportError:
    print("Installing Ibis, this is a one-time setup...")
    %pip install -q "ibis-framework[postgres]"
    import ibis
    print("Ibis installed. If the next cell fails, restart the kernel and run it again.")
````

With Ibis, you connect to a database and get a table object you can filter and select on, much like working on a dataframe. Here is the same kind of query you wrote above, but in Python:

````python
from rel_db_learning import get_ibis_chinook

ibis_chinook = get_ibis_chinook()
artist = ibis_chinook.table("artist")

# Artists whose name starts with 'A'
result = artist.filter(artist.name.like("A%")).select("name")
result.execute()
````

The interesting part: Ibis does not run Python against the data. It
translates your Python into SQL and sends that to the database. You can see
the generated SQL:

````python
print(ibis.to_sql(result))
````

Notice it produced a SELECT statement, the same kind you have been writing by
hand. Ibis is a different way to express the same query.
